```python
A = np.array([[1, 2],
              [3, 1j]])
array_to_latex(A, prefix='A = ')
```

## Matrix addition

```python
A = np.array([[1, 0],
              [0, 1j]])
array_to_latex(A, prefix='A = ')
```

## Matrix multiplication

```python
A = np.array([[1, 0],
              [0, 1j]])
array_to_latex(A, prefix='A = ')
```

```python
B = np.array([[1, 1, 1j],
              [1, -1, 0]])
array_to_latex(B, prefix='B = ')
```

```python
array_to_latex(A@B, prefix='A@B = ')
```

## Matrix inverse

```python
from numpy.linalg import inv
A = np.array([[1., 2.], [3., 4.]])
array_to_latex(A, prefix='A =')
```

```python
array_to_latex(inv(A), prefix='A^{-1} = ')
```

## Kronecker product or tensor product of matrices or vectors

```python
A = np.array([[1,2],
              [3, 4]])
array_to_latex(A, prefix='A =')
```

```python
B = np.array([[0, 5],
              [6, 7]])
array_to_latex(B, prefix='B =')
```

```python
C = np.kron(A, B)
array_to_latex(C, prefix=r'A \otimes B =')
```
